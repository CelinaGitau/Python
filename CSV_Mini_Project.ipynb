```python
from google.colab import files
grades= files.upload()
```

```python
#read csv file
import pandas as pd
grades = pd.read_csv ("grades.csv")
grades
```

```python
# Check column names to check for exact match
print(grades.columns)
grades[" Grade"] = grades[" Grade"].astype(float)
grades
```

```python
print("\nAfter Cleaning (Grades as float):")
print(grades.dtypes)
```

```python
#summarise the data
avg_grade = grades[" Grade"].mean()
print("Average:", avg_grade)
```

```python
avg_grade = grades[" Grade"].mean()
min_grade = grades[" Grade"].min()
max_grade = grades[" Grade"].max()
num_students = grades[" Grade"].count()

print("Number of Students:", num_students)
print("Average:", avg_grade)
print("Minimum:", min_grade)
print("Maximum:", max_grade)
print ("Number of Students:", num_students)
```

```python
#Ouput the summary to a new CSV
summary_df = pd.DataFrame({
    "Average": [avg_grade],
    "Minimum": [min_grade],
    "Maximum": [max_grade],
    "Number of Students": [num_students]
})

#saves the DataFframe as a CSV file
summary_df.to_csv("summary.csv", index=False) # this removes the row numbers
```

```python
#prints a new line, lets you know the program finished successfully
print("\n Summary CSV file created successfully")
```
